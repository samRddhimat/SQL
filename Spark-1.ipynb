```scala
%spark.pyspark
# Create an Iceberg database and table
spark.sql("CREATE DATABASE IF NOT EXISTS iceberg.my_db")
spark.sql("""
    CREATE TABLE IF NOT EXISTS iceberg.my_db.users (
        id INT,
        name STRING
    ) USING iceberg
""")

# Insert data and query it
spark.sql("INSERT INTO iceberg.my_db.users VALUES (1, 'Alice'), (2, 'Bob')")
df = spark.sql("SELECT * FROM iceberg.my_db.users")
df.show()
```

```scala
%spark.pyspark
spark.sql("CREATE DATABASE IF NOT EXISTS iceberg.my_db")
spark.sql("""
	CREATE TABLE empsal (
	EmployeeID	INTEGER,
	Name	STRING,
	Department	STRING,
	Experienceyrs	INTEGER,
	Education	STRING,
	Age	INTEGER,
	Gender	STRING,
	City	STRING,
	Monthly_Salary	DECIMAL(10,2)
) USING iceberg
""")
```

```scala
%spark.pyspark

spark.sql("USE iceberg.my_db;")
# spark.sql("""
# 	CREATE TABLE empsal (
# 	EmployeeID	INTEGER,
# 	Name	STRING,
# 	Department	STRING,
# 	Experienceyrs	INTEGER,
# 	Education	STRING,
# 	Age	INTEGER,
# 	Gender	STRING,
# 	City	STRING,
# 	Monthly_Salary	DECIMAL(10,2)
# ) USING iceberg
# """)

df = spark.sql("""

select * from
 (select name, department,monthly_salary, 
avg(monthly_salary) over(partition by department) as dept_avg_salary from empsal
) e
where monthly_salary > dept_avg_salary 

""")


#df = spark.sql("SELECT * FROM iceberg.my_db.empsal")
df.show()

```

```scala
%spark.pyspark

df = spark.sql("""

WITH deptavg as 
(select department,avg(monthly_salary) avgms from empsal group by department)

select a.name,a.department,a.monthly_salary, d.avgms as dept_avg  from empsal a
inner join deptavg d on a.department = d.department
where a. monthly_salary > d.avgms order by a.department

""")


#df = spark.sql("SELECT * FROM iceberg.my_db.empsal")
df.show()

```
